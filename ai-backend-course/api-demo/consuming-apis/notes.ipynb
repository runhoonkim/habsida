```python
# requests_get.py
import requests

response = requests.get("https://jsonplaceholder.typicode.com/users/1")

print(response.status_code)        # 200
print(response.headers["Content-Type"])  # application/json; charset=utf-8
data = response.json()             # parse the JSON body into a Python dict
print(data["name"], data["email"])
```

```python
import json
print(json.dumps(data, indent=2))
```

```python
print(json.dumps(dict(response.headers), indent=2))
```

```python
# requests_param.py
import requests

response = requests.get(
    "https://jsonplaceholder.typicode.com/posts",
    params={"userId": 1,
            'id':1},
)
print(f"url: {response.url}") # url: https://jsonplaceholder.typicode.com/posts?userId=1&id=1
posts = response.json()
print(f"User 1 has {len(posts)} posts")
```

```python
# requests_posts.py
import requests

new_post = {"title": "First order", "body": "Two units", "userId": 1}
response = requests.post(
    "https://jsonplaceholder.typicode.com/posts",
    json=new_post,
)
print(response.status_code)   # 201 Created
print(response.json()["id"])  # the id the server assigned
```

```python
headers = {"Authorization": "Bearer demo-token"}
response = requests.get("https://jsonplaceholder.typicode.com/users/999", headers=headers)

if response.status_code == 404:
    print("User not found")
else:
    response.raise_for_status()   # raise an exception on any 4xx/5xx
    print(response.json())
```
