| Piece | Purpose |
|---|---|
| `@app.get(...)` | Registers the function as a GET endpoint. |
| `"/orders/{order_id}"` | Defines the URL pattern. |
| `response_model=OrderOut` | Defines the response schema. |
| `order_id: int` | Tells FastAPI to parse the URL value as an integer. |
| `Depends(get_db)` | Tells FastAPI how to supply the `db` argument. |
